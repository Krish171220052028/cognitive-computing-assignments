## Q2. Basic NumPy array questions

**(a)** Reverse the array `arr = np.array([1, 2, 3, 6, 4, 5])`.

**(b)** Find the most frequent value and its index/indices in:
- `x = np.array([1, 2, 3, 4, 5, 1, 2, 1, 1, 1])`
- `y = np.array([1, 1, 1, 2, 3, 4, 2, 4, 3, 3])`

````python
import numpy as np
from collections import Counter

arr = np.array([1, 2, 3, 6, 4, 5])
print("Reversed array:", arr[::-1])

x = np.array([1, 2, 3, 4, 5, 1, 2, 1, 1, 1])
y = np.array([1, 1, 1, 2, 3, 4, 2, 4, 3, 3])

def most_frequent_and_indices(a):
    counts = Counter(a.tolist())
    highest = max(counts.values())
    modes = [value for value, count in counts.items() if count == highest]
    indices = {value: np.where(a == value)[0].tolist() for value in modes}
    return modes, highest, indices

for label, a in [("x", x), ("y", y)]:
    modes, frequency, indices = most_frequent_and_indices(a)
    print(f"{label}: most frequent value(s) = {modes}, frequency = {frequency}")
    print(f"{label}: index/indices = {indices}")
````

**Expected output**
```text
Reversed array: [5 4 6 3 2 1]
x: most frequent value(s) = [1], frequency = 5
x: index/indices = {1: [0, 5, 7, 8, 9]}
y: most frequent value(s) = [1, 3], frequency = 3
y: index/indices = {1: [0, 1, 2], 3: [4, 8, 9]}
```

**Note:** For `y`, there is a tie: both 1 and 3 appear three times. The code returns all tied most-frequent values and their zero-based indices.

## Q3. Access elements in a 2-D array

For the matrix
```python
arr = np.array([[10, 20, 30],
                [40, 50, 60],
                [70, 80, 90]])
```
- (a) Access the first row, second column.
- (b) Access the third row, first column.

## Q4. Create a 1-D array using `linspace()`

Create a 1-D NumPy array named `Krish_Marwaha` containing 25 evenly spaced numbers from 10 to 100. Print its dimensions, shape, total elements, element data type, and bytes consumed. Find the transpose using `reshape()`. Can this be done using the `.T` attribute?

````python
import numpy as np

Krish_Marwaha = np.linspace(10, 100, 25)

print("Array:", Krish_Marwaha)
print("Dimensions:", Krish_Marwaha.ndim)
print("Shape:", Krish_Marwaha.shape)
print("Total elements:", Krish_Marwaha.size)
print("Data type:", Krish_Marwaha.dtype)
print("Bytes consumed:", Krish_Marwaha.nbytes)

column_array = Krish_Marwaha.reshape(25, 1)
print("Transpose using reshape (25 x 1):\n", column_array)
print("Transpose using .T (same result):\n", Krish_Marwaha.T)
````

**Expected output**
```text
Array: [ 10.    13.75  17.5   21.25  25.    28.75  32.5   36.25  40.
  43.75  47.5   51.25  55.    58.75  62.5   66.25  70.    73.75
  77.5   81.25  85.    88.75  92.5   96.25 100.  ]
Dimensions: 1
Shape: (25,)
Total elements: 25
Data type: float64
Bytes consumed: 200
```

The `reshape(25, 1)` operation changes the array into a 25-row, 1-column matrix. The `.T` attribute on a 1-D array does **not** change its shape; it remains `(25,)`. To obtain a column vector, use `reshape(25, 1)` or `[:, None]`.

## Q5. Create and analyze a 2-D array

Create a 2-D array named `ucs420_krish_marwaha` with three rows and four columns using the values `10, 20, 30, 40, 50, 60, 70, 80, 90, 15, 20, 35`. Compute mean, median, maximum, minimum, and unique elements. Reshape it to four rows and three columns as `reshaped_ucs420_krish_marwaha`. Resize it to two rows and three columns as `resized_ucs420_krish_marwaha`.

````python
import numpy as np

ucs420_krish_marwaha = np.array([
    [10, 20, 30, 40],
    [50, 60, 70, 80],
    [90, 15, 20, 35]
])

print("Original array:\n", ucs420_krish_marwaha)
print("Mean:", np.mean(ucs420_krish_marwaha))
print("Median:", np.median(ucs420_krish_marwaha))
print("Maximum:", np.max(ucs420_krish_marwaha))
print("Minimum:", np.min(ucs420_krish_marwaha))
print("Unique elements:", np.unique(ucs420_krish_marwaha))

reshaped_ucs420_krish_marwaha = ucs420_krish_marwaha.reshape(4, 3)
print("Reshaped array (4 x 3):\n", reshaped_ucs420_krish_marwaha)

resized_ucs420_krish_marwaha = np.resize(ucs420_krish_marwaha, (2, 3))
print("Resized array (2 x 3):\n", resized_ucs420_krish_marwaha)
````

**Expected output**
```text

 [[10 20 30 40]
 [50 60 70 80]
 [90 15 20 35]]
Mean: 43.333333333333336
Median: 37.5
Maximum: 90
Minimum: 10
Unique elements: [10 15 20 30 35 40 50 60 70 80 90]
Reshaped array (4 x 3):
 [[10 20 30]
 [40 50 60]
 [70 80 90]
 [15 20 35]]
Resized array (2 x 3):
 [[10 20 30]
 [40 50 60]]
```

`reshape()` changes the shape while keeping all 12 elements. `np.resize()` returns an array with the requested shape; when the new shape contains fewer elements, the remaining values are omitted.
