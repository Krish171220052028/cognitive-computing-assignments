## Q1. Sensor readings and vectorization

Consider the sensor readings:

`temperature = np.array([25, 28, 31, 35, 38, 27, 33, 40])`

A sensor has an average measurement error of +2°C.

Perform the following without using a `for` loop:

a. Add 2°C to every reading using vectorization.  
b. Convert all corrected temperatures from Celsius to Fahrenheit using `F = (9/5)C + 32`.  
c. Identify readings greater than 32°C using Boolean indexing.  
d. Count how many readings exceed 32°C.  
e. Explain how vectorization and Boolean indexing are more efficient than explicitly iterating through every value.

````python
import numpy as np

temperature = np.array([25, 28, 31, 35, 38, 27, 33, 40])

# a. Correct readings using vectorization
corrected = temperature + 2

# b. Convert corrected Celsius values to Fahrenheit
fahrenheit = (9 / 5) * corrected + 32

# c. Boolean indexing for readings greater than 32°C
above_32 = corrected[corrected > 32]

# d. Count readings greater than 32°C
count_above_32 = np.sum(corrected > 32)

print("Corrected temperatures:", corrected)
print("Temperatures in Fahrenheit:", fahrenheit)
print("Readings above 32 C:", above_32)
print("Count above 32 C:", count_above_32)
````

**Expected output**
```text
Corrected temperatures: [27 30 33 37 40 29 35 42]
Temperatures in Fahrenheit: [80.6 86.  91.4 98.6 104.  84.2 95.  107.6]
Readings above 32 C: [33 37 40 35 42]
Count above 32 C: 5
```

**(e) Explanation:** Vectorization applies an operation to an entire NumPy array without an explicit Python loop. Boolean indexing selects only values satisfying a condition. Both make numerical code shorter and generally more efficient.

## Q2. Daily steps matrix

The following matrix records daily steps for four users over three days:

```python
steps = np.array([
    [5000, 6200, 7100],
    [8000, 7500, 9000],
    [4500, 5100, 4800],
    [9000, 8500, 9500]
])
```

Calculate the total, mean, maximum, minimum, total steps for each day using `axis=0`, total steps for each user using `axis=1`, and the user/day position of the maximum using `argmax()`. 

````python
import numpy as np

steps = np.array([
    [5000, 6200, 7100],
    [8000, 7500, 9000],
    [4500, 5100, 4800],
    [9000, 8500, 9500]
])

print("Total steps:", np.sum(steps))
print("Mean steps:", np.mean(steps))
print("Maximum steps:", np.max(steps))
print("Minimum steps:", np.min(steps))
print("Total per day:", np.sum(steps, axis=0))
print("Total per user:", np.sum(steps, axis=1))
print("Maximum position (row, column):", np.unravel_index(np.argmax(steps), steps.shape))
````

The maximum is at row index 3 and column index 2 (fourth user, third day), because NumPy indexing starts at zero.

## Q3. NumPy slicing, copying, reshaping, and array properties

Write a program to:
- Create `original = np.array([1, 2, 3, 4, 5, 6])`.
- Slice indices 1 to 4 into `subset`, change its first element to 999, and display both arrays.
- Create another slice using `.copy()`, change its first element to 500, and display both arrays.
- Create a 3 × 4 matrix using `np.arange()` and `reshape()`.
- Extract the first row, last row, second column, and rows 1–2 with columns 2–3.
- Create flattened versions using `flatten()` and `ravel()`, modify each, and observe whether the original matrix changes.
- Display `shape`, `ndim`, `size`, and `dtype`.

````python
import numpy as np

original = np.array([1, 2, 3, 4, 5, 6])
subset = original[1:5]
subset[0] = 999
print("After changing subset:")
print("original:", original)
print("subset:", subset)

original2 = np.array([1, 2, 3, 4, 5, 6])
copied = original2[1:5].copy()
copied[0] = 500
print("\nAfter changing copied array:")
print("original:", original2)
print("copied:", copied)

matrix = np.arange(1, 13).reshape(3, 4)
print("\nMatrix:\n", matrix)
print("First row:", matrix[0, :])
print("Last row:", matrix[-1, :])
print("Second column:", matrix[:, 1])
print("Rows 1-2 and columns 2-3:", matrix[1:3, 2:4])

raveled = matrix.ravel()
raveled[0] = 100
print("\nAfter changing ravel result, matrix:\n", matrix)

flattened = matrix.flatten()
flattened[1] = 200
print("After changing flatten result, matrix:\n", matrix)
print("flattened:", flattened)

print("shape:", matrix.shape)
print("ndim:", matrix.ndim)
print("size:", matrix.size)
print("dtype:", matrix.dtype)
````

## Q4. Ordinary Least Squares for assistance score prediction

A cognitive-assistive system predicts an assistance score using sleep hours, activity level, and stress level.

```python
X = np.array([
    [6, 70, 3],
    [5, 50, 6],
    [8, 80, 2],
    [4, 30, 8]
])
y = np.array([40, 65, 30, 85])
```

a. Display the shape and dimensions of `X`.  
b. Calculate `X.T` and explain the transpose.  
c. Calculate `X.T @ X`.  
d. Calculate the inverse using `np.linalg.inv()` where appropriate.  
e. Implement the OLS equation `β = (X.T @ X)^(-1) @ X.T @ y`.  
f. Explain what the coefficients represent.  
g. Predict the score for `new_user = np.array([5, 40, 7])`.

````python
import numpy as np

X = np.array([
    [6, 70, 3],
    [5, 50, 6],
    [8, 80, 2],
    [4, 30, 8]
], dtype=float)

y = np.array([40, 65, 30, 85], dtype=float)

print("Shape of X:", X.shape)
print("Dimensions of X:", X.ndim)
print("X transpose:\n", X.T)

xtx = X.T @ X
print("X.T @ X:\n", xtx)

try:
    xtx_inv = np.linalg.inv(xtx)
    beta = xtx_inv @ X.T @ y
    print("Inverse of X.T @ X:\n", xtx_inv)
except np.linalg.LinAlgError:
    print("X.T @ X is singular; using the pseudoinverse.")
    beta = np.linalg.pinv(X) @ y

print("Coefficients:", beta)
new_user = np.array([5, 40, 7], dtype=float)
print("Predicted assistance score:", new_user @ beta)
````

**Interpretation:** Each coefficient estimates the change in predicted assistance score for a one-unit increase in its feature, holding the other features constant. The features are sleep hours, activity level, and stress level. The equation specified in the assignment has no intercept, so this model should be interpreted accordingly.

**Numerical note:** The code checks whether the inverse exists. If `X.T @ X` is singular, it uses a pseudoinverse rather than reporting a nonexistent inverse.
