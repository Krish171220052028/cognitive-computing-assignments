## Question 1 — Lists

Extract the individual digits of the roll number and multiply each digit by 10.

```python
# Q1: Create list L from the digits of the roll number
roll_no = 1024170223
digits = [int(d) for d in str(roll_no)]
L = [digit * 10 for digit in digits]

print("L =", L)
```

```python
# ii. Add two numbers: append() adds at the end, while insert() adds at a chosen position.
L.append(250)
print("After append(250):", L)

L.insert(2, 350)
print("After insert(2, 350):", L)
```

```python
# iii. Remove two elements.
# remove() deletes the first matching value; pop() deletes and returns an element by index.
L.remove(20)
print("After remove(20):", L)

removed = L.pop(1)
print("After pop(1):", L)
print("Element removed by pop():", removed)
```

```python
# iv. Sort in ascending and descending order.
L.sort()
print("Ascending:", L)

L.sort(reverse=True)
print("Descending:", L)
```

```python
# v. Slicing: first three and last three elements.
print("First three elements:", L[:3])
print("Last three elements:", L[-3:])
```

```python
# vi. List comprehension: elements greater than the average.
average = sum(L) / len(L)
greater_than_average = [x for x in L if x > average]

print("Average =", average)
print("Elements greater than average:", greater_than_average)
```

## Question 2 — Tuples

Create `scores` using the first 8 values from the original list L in Q1.

```python
# Q2: The first 8 values from the original digit-based list
original_L = [digit * 10 for digit in digits]
scores = tuple(original_L[:8])

print("scores =", scores)
```

```python
# i. Highest score and its index; lowest score and its frequency.
highest = max(scores)
highest_index = scores.index(highest)

lowest = min(scores)
lowest_count = scores.count(lowest)

print("Highest score:", highest)
print("Index of highest score:", highest_index)
print("Lowest score:", lowest)
print("Number of times lowest score appears:", lowest_count)
```

```python
# ii. Reverse the tuple and return it as a list.
# Tuples are immutable, so they cannot be reversed in place like lists.
reversed_scores = list(reversed(scores))
print("Reversed tuple as a list:", reversed_scores)
```

```python
# iii. Search for a score entered by the user.
user_score = int(input("Enter a score to search in scores: "))

if user_score in scores:
    print("First occurrence index:", scores.index(user_score))
else:
    print("Score is not present in the tuple.")
```

```python
# iv. Attempt to modify a tuple element.
# A tuple does not allow item assignment because it is immutable, unlike a list.
try:
    scores[0] = 100
except TypeError as e:
    print("Python error:", e)
    print("Explanation: Tuples are immutable, whereas list elements can be changed.")
```

```python
# v. Unpack into first score, second score and all remaining scores.
first_score, second_score, *remaining_scores = scores

print("First score:", first_score)
print("Second score:", second_score)
print("Remaining scores:", remaining_scores)
```

## Question 3 — Random Numbers

Use the roll number as the random seed so the generated list is reproducible and unique to the roll number.

```python
# Q3(i): Generate 100 reproducible random integers from 100 to 900 inclusive.
import random

random.seed(1024170223)
random_numbers = [random.randint(100, 900) for _ in range(100)]

print("Random list:")
print(random_numbers)
```

```python
# ii. Count and print all odd numbers.
odd_numbers = [x for x in random_numbers if x % 2 != 0]
print("Odd numbers:", odd_numbers)
print("Count of odd numbers:", len(odd_numbers))
```

```python
# iii. Count and print all even numbers.
even_numbers = [x for x in random_numbers if x % 2 == 0]
print("Even numbers:", even_numbers)
print("Count of even numbers:", len(even_numbers))
```

```python
# iv. Count and print prime numbers, and build the prime list using list comprehension.
def is_prime(n):
    if n < 2:
        return False
    for i in range(2, int(n ** 0.5) + 1):
        if n % i == 0:
            return False
    return True

prime_numbers = [x for x in random_numbers if is_prime(x)]

print("Prime numbers:", prime_numbers)
print("Count of prime numbers:", len(prime_numbers))
```

```python
# v. Print the most frequent number and its frequency.
from collections import Counter

frequency = Counter(random_numbers)
most_frequent_number, highest_frequency = frequency.most_common(1)[0]

print("Most frequent number:", most_frequent_number)
print("Frequency:", highest_frequency)
```

## Question 4 — Sets

Use the same 8 roll-number digits from Q1 before multiplying them by 10.

```python
# Q4: Create sets A and B from the first 8 digits.
first_8_digits = digits[:8]

A = {digit * 7 for digit in first_8_digits}
B = {digit * 9 for digit in first_8_digits}

print("A =", A)
print("B =", B)
```

```python
# vi. Union: all unique values present in A or B.
union_AB = A.union(B)
print("A union B =", union_AB)
```

```python
# vii. Intersection: values common to both sets.
intersection_AB = A.intersection(B)
print("A intersection B =", intersection_AB)
```

```python
# viii. Differences in both directions.
A_minus_B = A.difference(B)
B_minus_A = B.difference(A)

print("A - B =", A_minus_B)
print("B - A =", B_minus_A)

# difference() gives one directional difference; symmetric_difference() gives values in either set but not both.
```

```python
# ix. Symmetric difference: values present in exactly one of the two sets.
symmetric_AB = A.symmetric_difference(B)
print("Symmetric difference =", symmetric_AB)
```

```python
# x. Subset and superset checks.
print("Is A a subset of B?", A.issubset(B))
print("Is B a superset of A?", B.issuperset(A))
```

```python
# xi. Safely discard a user-entered value from A.
X = int(input("Enter a value to discard from A: "))
A.discard(X)

print("A after discard:", A)

# discard() does not raise an error if X is absent; remove() would raise KeyError in that case.
```

## Question 5 — Dictionaries

```python
# Given dictionary
my_dict = {
    "name": "Krish Marwaha",
    "roll_no": "1024170223",
    "branch": "CSE",
    "age": 20,
    "city": "Amritsar"
}

print("Original dictionary:")
print(my_dict)
```

```python
# i. Rename "city" to "location" without recreating the dictionary by hand.
my_dict["location"] = my_dict.pop("city")
print("After renaming city to location:", my_dict)
```

```python
# ii. Add CGPA.
my_dict["cgpa"] = 8.47
print("After adding CGPA:", my_dict)
```

```python
# iii. Increase age by 1.
my_dict["age"] += 1
print("After increasing age by 1:", my_dict)
```

```python
# iv. Delete "branch" using pop() and del on separate copies.
dict_pop = my_dict.copy()
removed_branch = dict_pop.pop("branch")
print("Using pop():", dict_pop)
print("Value returned by pop():", removed_branch)

dict_del = my_dict.copy()
del dict_del["branch"]
print("Using del:", dict_del)

# pop() removes and returns the value; del removes the key-value pair but does not return the value.
```

```python
# v. Iterate using .items() and print each key-value pair.
for key, value in my_dict.items():
    print(f"{key} → {value}")
```

```python
# vi. Safely check for a missing "email" key.
if "email" in my_dict:
    print("Email:", my_dict["email"])
else:
    print("Email key is not present in the dictionary.")
```

```python
# vii. Create a fictional friend's dictionary and merge it with the original.
friend_dict = {
    "name": "Rahul Sharma",
    "roll_no": "1024170999",
    "branch": "CSE",
    "age": 20,
    "city": "Delhi"
}

merged_dict = {**my_dict, **friend_dict}
print("Friend dictionary:", friend_dict)
print("Merged dictionary:", merged_dict)

# When the same key exists in both dictionaries, values from the second dictionary win.
```

```python
# viii. Keep only key-value pairs whose values are strings.
string_values_dict = {
    key: value for key, value in my_dict.items()
    if isinstance(value, str)
}

print("Dictionary containing only string values:", string_values_dict)
```
